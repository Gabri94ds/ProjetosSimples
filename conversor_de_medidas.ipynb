```python
def menu_principal():

  unidades_medida = {1: 'Tempo', 2: 'Massa', 3: 'Distância', 4: 'Área', 5: 'Volume', 6: 'litro'}

  print('Selecione abaixo qual medida você quer fazer a conversão:')
  for n, med in unidades_medida.items():
    print(f'{n}: {med}')
  return int(input('Digite a opcao: '))

def opcao1():

  opcao = 0

  unidades_tempo = {1: 'horas', 2: 'minutos', 3: 'segundos'}

  para_segundos = {'horas': 3600, 'minutos': 60, 'segundos': 1}

  while opcao != 4:
    print('\nSelecione abaixo qual unidade de medida inicial')
    for num, nome in unidades_tempo.items():
      print(f'{num} - {nome.capitalize()}')
    print('4 - Retornar')

    try:
      opcao = int(input('Digite a opcao: '))
    except ValueError:
      print('Por favor, digite apenas número inteiro')
      continue

    if opcao in unidades_tempo:
      unidade_origem = unidades_tempo[opcao]

      num = float(input(f'Digite o valor em {unidade_origem}: '))
      print(f'Você quer converter {num} {unidade_origem} para:')
      opcoes_destino = {k: v for k, v in unidades_tempo.items() if k != opcao}
      for num_op, nome in opcoes_destino.items():
                print(f'{num_op} - {nome.capitalize()}')
      try:
        conversao = int(input('Digite a opcao de conversão: '))
      except ValueError:
        print('Por favor, digite apenas número inteiro')
        continue

      if conversao in opcoes_destino:
        unidade_destino = opcoes_destino[conversao]
        resultado = num * para_segundos[unidade_origem] / para_segundos[unidade_destino]
        print(f'{num} {unidade_origem} = {resultado} {unidade_destino}')

      else:
        print('Opção inválida')

    else:
      if opcao != 4:
        print('Opção inválida')

def opcao2():

  opcao = 0

  unidade_massa = {1:'tonelada', 2:'kilograma', 3: 'grama'}
  para_grama = {'tonelada': 1000000, 'kilograma': 1000, 'grama': 1}

  while opcao != 4:
    print('\nSelecione abaixo qual unidade de medida inicial')
    for num, nome in unidade_massa.items():
      print(f'{num} - {nome.capitalize()}')
    print('4 - Retornar')

    try:
      opcao = int(input('Digite a opção: '))
    except ValueError:
      print('Por favor, digite apenas número inteiro')
      continue

    if opcao in unidade_massa:
      unidade_origem = unidade_massa[opcao]

      num = float(input(f'Digite o valor em {unidade_origem}: '))
      print(f'Você quer converter {num} {unidade_origem} para:')

      opcoes_destino = {k: v for k, v in unidade_massa.items() if k != opcao}
      for num_op, nome in opcoes_destino.items():
                print(f'{num_op} - {nome.capitalize()}')
      try:
        conversao = int(input('Digite a opcao de conversão: '))
      except ValueError:
        print('Por favor, digite apenas número inteiro')
        continue

      if conversao in opcoes_destino:
        unidade_destino = opcoes_destino[conversao]
        resultado = num * para_grama[unidade_origem] / para_grama[unidade_destino]
        print(f'{num} {unidade_origem} = {resultado} {unidade_destino}')

      else:
        print('Opcao inválida')

    else:
      if opcao != 4:
        print('Opção inválida')

def opcao3():

  opcao = 0

  unidade_distancia = {1:'kilometro', 2:'metro'}
  para_metro = {'kilometro': 1000, 'metro': 1}

  while opcao != 3:
    print('\nSelecione abaixo qual unidade de medida inicial')
    for num, nome in unidade_distancia.items():
      print(f'{num} - {nome.capitalize()}')
    print('3 - Retornar')

    try:
      opcao = int(input('Digite a opção: '))
    except ValueError:
      print('Por favor, digite apenas número inteiro')
      continue

    if opcao in unidade_distancia:
      unidade_origem = unidade_distancia[opcao]
      opcoes_destino = {k: v for k, v in unidade_distancia.items() if k != opcao}
      unidade_destino = list(opcoes_destino.values())[0]

      num = float(input(f'Digite o valor em {unidade_origem}: '))
      print(f'Você quer converter {num} {unidade_origem} para {unidade_destino}')

      resultado = num * para_metro[unidade_origem] / para_metro[unidade_destino]
      print(f'{num} {unidade_origem} = {resultado} {unidade_destino}')

    else:
      if opcao != 3:
        print('Opção inválida')

def opcao4():

  opcao = 0

  unidade_area = {1:'kilometro²', 2:'metro²'}
  para_metro = {'kilometro²': 1000000, 'metro²': 1}

  while opcao != 3:
    print('\nSelecione abaixo qual unidade de medida inicial')
    for num, nome in unidade_distancia.items():
      print(f'{num} - {nome.capitalize()}')
    print('3 - Retornar')

    try:
      opcao = int(input('Digite a opção: '))
    except ValueError:
      print('Por favor, digite apenas número inteiro')
      continue

    if opcao in unidade_area:
      unidade_origem = unidade_area[opcao]
      opcoes_destino = {k: v for k, v in unidade_area.items() if k != opcao}
      unidade_destino = list(opcoes_destino.values())[0]

      num = float(input(f'Digite o valor em {unidade_origem}: '))
      print(f'Você quer converter {num} {unidade_origem} para {unidade_destino}')

      resultado = num * para_metro[unidade_origem] / para_metro[unidade_destino]
      print(f'{num} {unidade_origem} = {resultado} {unidade_destino}')

    else:
      if opcao != 3:
        print('Opção inválida')

def opcao5():

  opcao = 0

  unidade_volume = {1:'kilometro³', 2:'metro³'}
  para_metro = {'kilometro³': 1000000000, 'metro³': 1}

  while opcao != 3:
    print('\nSelecione abaixo qual unidade de medida inicial')
    for num, nome in unidade_distancia.items():
      print(f'{num} - {nome.capitalize()}')
    print('3 - Retornar')

    try:
      opcao = int(input('Digite a opção: '))
    except ValueError:
      print('Por favor, digite apenas número inteiro')
      continue

    if opcao in unidade_volume:
      unidade_origem = unidade_volume[opcao]
      opcoes_destino = {k: v for k, v in unidade_volume.items() if k != opcao}
      unidade_destino = list(opcoes_destino.values())[0]

      num = float(input(f'Digite o valor em {unidade_origem}: '))
      print(f'Você quer converter {num} {unidade_origem} para {unidade_destino}')

      resultado = num * para_metro[unidade_origem] / para_metro[unidade_destino]
      print(f'{num} {unidade_origem} = {resultado} {unidade_destino}')

    else:
      if opcao != 3:
        print('Opção inválida')

def opcao6():

  opcao = 0

  unidade_litro = {1:'litro', 2:'mililitro'}
  para_litro = {'litro': 1000, 'mililitro': 1}

  while opcao != 3:
    print('\nSelecione abaixo qual unidade de medida inicial')
    for num, nome in unidade_litro.items():
      print(f'{num} - {nome.capitalize()}')
    print('3 - Retornar')

    try:
      opcao = int(input('Digite a opção: '))
    except ValueError:
      print('Por favor, digite apenas número inteiro')
      continue

    if opcao in unidade_litro:
      unidade_origem = unidade_litro[opcao]
      opcoes_destino = {k: v for k, v in unidade_litro.items() if k != opcao}
      unidade_destino = list(opcoes_destino.values())[0]

      num = float(input(f'Digite o valor em {unidade_origem}: '))
      print(f'Você quer converter {num} {unidade_origem} para {unidade_destino}')

      resultado = num * para_litro[unidade_origem] / para_litro[unidade_destino]
      print(f'{num} {unidade_origem} = {resultado} {unidade_destino}')

    else:
      if opcao != 3:
        print('Opção inválida')


def main():
  opcao = menu_principal()

  if opcao in range(1,7):
    if opcao == 1:
      opcao1()

    if opcao == 2:
      opcao2()

    if opcao == 3:
      opcao3()

    if opcao == 4:
      opcao4()

    if opcao == 5:
      opcao5()

    if opcao == 6:
      opcao6()

  else:
    print('Opção inválida!')


main()
```
